```python
# Perceptron em Python

import numpy as np

class Perceptron:
    def __init__(self, taxa_aprendizagem=0.01, n_iteracoes=50):
        self.taxa_aprendizagem = taxa_aprendizagem
        self.n_iteracoes = n_iteracoes

    def fit(self, X, y):
        """Ajusta o perceptron aos dados de treinamento."""
        # Inicializa os pesos e o viés com zeros
        self.pesos = np.zeros(X.shape[1])
        self.bias = 0

        # Converte os rótulos para -1 ou 1 (necessário para o algoritmo do perceptron)
        y_transformado = np.where(y == 0, -1, 1)

        # Itera sobre as épocas (iterações)
        for _ in range(self.n_iteracoes):
            # Itera sobre cada amostra de treinamento
            for xi, target in zip(X, y_transformado):
                # Calcula a atualização do peso
                # (target - self.predict(xi)) é 1 se houver erro de classificação
                atualizacao = self.taxa_aprendizagem * (target - self.prever(xi))
                self.pesos += atualizacao * xi
                self.bias += atualizacao

    def net_input(self, X):
        """Calcula a entrada líquida."""
        return np.dot(X, self.pesos) + self.bias

    def prever(self, X):
        """Retorna 1 se a entrada líquida for >= 0, caso contrário retorna -1."""
        # A função de ativação degrau é usada aqui
        return np.where(self.net_input(X) >= 0.0, 1, -1)

# Exemplo de uso:

# Dados de exemplo (similar a uma porta AND)
X_treino = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
# Rótulos (0 para falso, 1 para verdadeiro - será convertido para -1 e 1 internamente)
y_treino = np.array([0, 0, 0, 1])

# Cria uma instância do Perceptron
perceptron = Perceptron(taxa_aprendizagem=0.1, n_iteracoes=10)

# Treina o perceptron
perceptron.fit(X_treino, y_treino)

# Testando as previsões
print("Previsões:")
print(f"Para [0, 0]: {perceptron.prever(np.array([0, 0]))}")
print(f"Para [0, 1]: {perceptron.prever(np.array([0, 1]))}")
print(f"Para [1, 0]: {perceptron.prever(np.array([1, 0]))}")
print(f"Para [1, 1]: {perceptron.prever(np.array([1, 1]))}")

# Note que o Perceptron só pode classificar conjuntos de dados linearmente separáveis.
# Para problemas não linearmente separáveis (como o XOR), ele não funcionará corretamente.
```

```python
# prompt: crie um algoritmo de classificação knn

from collections import Counter
from sklearn.model_selection import train_test_split
from sklearn import datasets

class KNN:
    def __init__(self, k=3):
        self.k = k

    def fit(self, X, y):
        self.X_treino = X
        self.y_treino = y

    def prever(self, X):
        previsoes = [self._prever(x) for x in X]
        return np.array(previsoes)

    def _prever(self, x):
        # Calcular distâncias entre x e todas as amostras de treinamento
        distancias = [self.distancia_euclidiana(x, x_treino) for x_treino in self.X_treino]

        # Obter os k vizinhos mais próximos (índices)
        indices_k_vizinhos = np.argsort(distancias)[:self.k]

        # Obter os rótulos dos k vizinhos mais próximos
        rotulos_k_vizinhos = [self.y_treino[i] for i in indices_k_vizinhos]

        # Determinar a classe mais comum (votação da maioria)
        mais_comum = Counter(rotulos_k_vizinhos).most_common(1)
        return mais_comum[0][0]

    def distancia_euclidiana(self, x1, x2):
        # Calcular a distância euclidiana
        return np.sqrt(np.sum((x1 - x2)**2))

# Exemplo de uso com o dataset Iris:
# Carregar o dataset Iris
iris = datasets.load_iris()
X, y = iris.data, iris.target

# Dividir os dados em conjuntos de treinamento e teste
X_treino, X_teste, y_treino, y_teste = train_test_split(X, y, test_size=0.2, random_state=1234)

# Criar uma instância do KNN
knn = KNN(k=3)

# Treinar o modelo
knn.fit(X_treino, y_treino)

# Fazer previsões no conjunto de teste
previsoes = knn.prever(X_teste)

# Avaliar a acurácia
acuracia = np.sum(previsoes == y_teste) / len(y_teste)
print(f"Acurácia do KNN: {acuracia:.4f}")

```
